```python
def is_integer(text):
    """Comprueba si una cadena representa un número entero utilizando try/except."""
    try:
        int(text)
        return True
    except ValueError:
        return False

def seconds_to_dhms(total_seconds):
    """Convierte una cantidad de segundos en días, horas, minutos y segundos."""
    dias = total_seconds // 86400
    resto = total_seconds % 86400
    horas = resto // 3600
    resto = resto % 3600
    minutos = resto // 60
    segundos = resto % 60
    return (dias, horas, minutos, segundos)

def main():
    """Función principal para probar el Ejercicio 1."""
    while True:
        entrada = input("Introduzca los segundos: ")
        if is_integer(entrada) and int(entrada) >= 0:
            segundos = int(entrada)
            break
        print("Entrada no válida. Introduzca un entero no negativo.")
    
    d, h, m, s = seconds_to_dhms(segundos)
    print(f"{segundos} segundos son {d} d, {h} h, {m} min y {s} s.")
```

```python
def is_leap_year(year):
    """Determina si un año es bisiesto con una única expresión booleana."""
    return (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)

def leap_years_between(start, end):
    """Devuelve la lista de años bisiestos en un rango dado."""
    bisiestos = []
    for y in range(start, end + 1):
        if is_leap_year(y):
            bisiestos.append(y)
    return bisiestos

def explain_leap_year(year):
    """Devuelve una explicación de por qué un año es o no bisiesto."""
    if is_leap_year(year):
        if year % 400 == 0:
            return f"{year} es bisiesto: es divisible entre 400."
        return f"{year} es bisiesto: es divisible entre 4 y no entre 100."
    else:
        if year % 4 == 0 and year % 100 == 0:
            return f"{year} no es bisiesto: es divisible entre 100 pero no entre 400."
        return f"{year} no es bisiesto: no es divisible entre 4."

def main_ex2():
    """Función principal para probar el Ejercicio 2."""
    entrada = input("Año: ")
    try:
        year = int(entrada)
        if year < 1582:
            print("El calendario gregoriano se estableció en 1582.")
        else:
            print(explain_leap_year(year))
    except ValueError:
        print("Entrada no válida.")
```

```python
def process_text(text: str = "This is a default text with Python and amazing words.", 
                 words_to_replace: list = ["Python", "amazing"]) -> tuple:
    """Procesa un texto reemplazando palabras indicadas por asteriscos."""
    text = text.lower().strip()
    words_in_text = text.split()
    
    words_to_replace_lower = [w.lower() for w in words_to_replace]
    replaced_count = 0
    processed_words = []
    
    for word in words_in_text:
        # Quitamos signos de puntuación básicos al final para la comprobación
        clean_word = word.strip(".,!?")
        if clean_word in words_to_replace_lower:
            # Reemplazamos conservando la puntuación
            new_word = ("*" * len(clean_word)) + word[len(clean_word):]
            processed_words.append(new_word)
            replaced_count += 1
        else:
            processed_words.append(word)
            
    processed_text = " ".join(processed_words)
    return processed_text, replaced_count

def main_ex3():
    """Función principal para probar el Ejercicio 3."""
    processed, count = process_text()
    print(f'Processed text: "{processed}"')
    print(f"Palabras reemplazadas: {count}")
```

```python
def shift_char(char: str, shift: int) -> str:
    """Desplaza un único carácter usando el cifrado César."""
    if not char.isalpha():
        return char
    
    base = ord('A') if char.isupper() else ord('a')
    # Aplicamos módulo 26 al desplazamiento
    new_char = chr(base + (ord(char) - base + shift) % 26)
    return new_char

def encrypt(text: str, shift: int) -> str:
    """Cifra un texto completo recorriéndolo carácter a carácter."""
    result = ""
    for char in text:
        result += shift_char(char, shift)
    return result

def decrypt(text: str, shift: int) -> str:
    """Descifra un texto reutilizando la función encrypt."""
    return encrypt(text, -shift)

def brute_force(cipher_text: str) -> list:
    """Devuelve los 26 posibles descifrados de un texto."""
    return [decrypt(cipher_text, i) for i in range(26)]

def main_ex4():
    """Función principal para probar el Ejercicio 4."""
    cifrado = encrypt("Hola, Mundo!", 3)
    print("Cifrado (shift=3):", cifrado)
    print("Descifrado:", decrypt(cifrado, 3))
    print("Fuerza bruta [3]:", brute_force(cifrado)[3])
```

```python
def normalize(text: str) -> str:
    """Normaliza un texto pasándolo a minúsculas, quitando tildes y caracteres no alfanuméricos."""
    text = text.lower()
    replacements = {'á': 'a', 'é': 'e', 'í': 'i', 'ó': 'o', 'ú': 'u', 'ü': 'u'}
    for k, v in replacements.items():
        text = text.replace(k, v)
    return "".join(c for c in text if c.isalnum())

def is_palindrome(text: str) -> bool:
    """Comprueba si un texto es palíndromo utilizando slicing [::-1]."""
    norm = normalize(text)
    return norm == norm[::-1]

def find_palindromes(sentences: list) -> list:
    """Filtra una lista de frases y devuelve solo los palíndromos."""
    return [s for s in sentences if is_palindrome(s)]

def palindrome_words(text: str) -> list:
    """Devuelve palabras palíndromas de al menos 3 letras dentro de una frase."""
    words = text.split()
    return [w for w in words if len(w) >= 3 and is_palindrome(w)]

def main_ex5():
    """Función principal para probar el Ejercicio 5."""
    sentences = [
        "Anita lava la tina",
        "Dábale arroz a la zorra el abad",
        "Esto no es un palíndromo",
        "¿Acaso hubo búhos acá?"
    ]
    print("Frases palíndromas:")
    for p in find_palindromes(sentences):
        print("-", p)
    print("Palabras palíndromas:", palindrome_words("Ana vio un oso en el ojo de Anita"))
```

```python
def is_prime(n: int) -> bool:
    """Comprueba si un número es primo."""
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

def prime_list(limit: int) -> list:
    """Genera una lista de números primos hasta un límite."""
    if limit < 2:
        return []
    return [i for i in range(2, limit + 1) if is_prime(i)]

def check_palindrome(primes: list) -> list:
    """Busca palíndromos en una lista de números primos."""
    return [p for p in primes if str(p) == str(p)[::-1]]

def categorize_prime(prime: int) -> str:
    """Clasifica un número primo por su tamaño."""
    if prime < 10:
        return "pequeño"
    elif prime < 100:
        return "mediano"
    else:
        return "grande"

def main_ex6():
    """Función principal para probar el Ejercicio 6."""
    primes = prime_list(150)
    palindromes = check_palindrome(primes)
    print("Primos palíndromos hasta 150:", palindromes)
    print("Clasificación del primo 101:", categorize_prime(101))
```

```python
def remove_duplicates(numbers: list) -> list:
    """Elimina duplicados conservando el orden original."""
    result = []
    for n in numbers:
        if n not in result:
            result.append(n)
    return result

def rotate_left(numbers: list, k: int) -> list:
    """Rota los elementos de la lista k posiciones a la izquierda."""
    if not numbers:
        return []
    k = k % len(numbers)
    return numbers[k:] + numbers[:k]

def split_even_odd(numbers: list) -> tuple:
    """Separa una lista en pares e impares."""
    pares = [n for n in numbers if n % 2 == 0]
    impares = [n for n in numbers if n % 2 != 0]
    return (pares, impares)

def merge_sorted(a: list, b: list) -> list:
    """Mezcla dos listas ordenadas recorriéndolas simultáneamente."""
    result = []
    i, j = 0, 0
    while i < len(a) and j < len(b):
        if a[i] < b[j]:
            result.append(a[i])
            i += 1
        else:
            result.append(b[j])
            j += 1
    result.extend(a[i:])
    result.extend(b[j:])
    return result

def second_largest(numbers: list):
    """Encuentra el segundo valor más grande de una lista."""
    if len(numbers) < 2:
        return None
    
    max1, max2 = float('-inf'), float('-inf')
    for n in numbers:
        if n > max1:
            max2 = max1
            max1 = n
        elif n > max2 and n != max1:
            max2 = n
            
    return max2 if max2 != float('-inf') else None

def main_ex7():
    """Función principal para probar el Ejercicio 7."""
    print("Duplicados eliminados:", remove_duplicates([3, 1, 3, 2, 1]))
    print("Rotación a la izquierda:", rotate_left([1, 2, 3, 4, 5], 2))
    print("Pares e impares:", split_even_odd([1, 2, 3, 4, 5, 6]))
    print("Fusión ordenada:", merge_sorted([1, 4, 9], [2, 3, 10, 11]))
    print("Segundo más grande:", second_largest([4, 9, 2, 9, 7]))
    print("Segundo más grande (iguales):", second_largest([5, 5]))
```

```python
def sum_all(nums: list) -> int:
    suma = 0
    for n in nums: suma += n
    return suma

def sum_pos_neg(nums: list) -> tuple:
    pos = sum([n for n in nums if n > 0])
    neg = sum([n for n in nums if n < 0])
    return pos, neg

def get_max_min(nums: list):
    if not nums: return None, None
    max_v, min_v = nums[0], nums[0]
    for n in nums:
        if n > max_v: max_v = n
        if n < min_v: min_v = n
    return max_v, min_v

def all_unique(nums: list) -> bool:
    vistos = []
    for n in nums:
        if n in vistos: return False
        vistos.append(n)
    return True

def compare_pos_neg(nums: list) -> str:
    pos = len([n for n in nums if n > 0])
    neg = len([n for n in nums if n < 0])
    return "Hay más positivos" if pos > neg else ("Hay más negativos" if neg > pos else "Igual cantidad de positivos y negativos")

def main_ex8():
    """Función principal para probar el Ejercicio 8."""
    # Para la prueba pedimos una lista
    lista = [1, -2, 3, 3, -5] 
    while True:
        print("\n1. Suma total | 2. Suma pos/neg | 3. Max/Min | 4. Únicos | 5. Comparar pos/neg | 6. Salir")
        opcion = input("Elige una opción: ")
        if opcion == '1': print("Suma:", sum_all(lista))
        elif opcion == '2': print("Positivos/Negativos:", sum_pos_neg(lista))
        elif opcion == '3':
            max_v, min_v = get_max_min(lista)
            print(f"Max: {max_v}, Min: {min_v}" if max_v is not None else "Lista vacía.")
        elif opcion == '4': print("Son únicos:", all_unique(lista))
        elif opcion == '5': print(compare_pos_neg(lista))
        elif opcion == '6' or opcion.lower() == 'salir': break
        else: print("Opción inválida.")
```

```python
def format_monomial(coef: float, exp: int) -> str:
    """Aplica las reglas de formateo para un único monomio."""
    if coef == 0:
        return ""
    
    # Manejo del coeficiente
    sign = "+" if coef > 0 else "-"
    abs_coef = abs(coef)
    
    if abs_coef == 1.0 and exp > 0:
        coef_str = sign
    else:
        coef_str = f"{sign}{abs_coef}" if exp > 0 or sign == "-" else f"+{abs_coef}"
        if sign == "-" and exp == 0:
            coef_str = f"-{abs_coef}"

    # Manejo de la x y el exponente
    if exp == 0:
        return coef_str
    elif exp == 1:
        return f"{coef_str}x"
    else:
        return f"{coef_str}x**{exp}"

def format_polynomial(coefs: list) -> str:
    """Formatea la lista de coeficientes como un polinomio completo."""
    if not coefs or all(c == 0 for c in coefs):
        return "0"
    
    terms = []
    for exp in range(len(coefs) - 1, -1, -1):
        monomial = format_monomial(coefs[exp], exp)
        if monomial:
            terms.append(monomial)
            
    if not terms:
        return "0"
        
    result = terms[0]
    if result.startswith("+"):
        result = result[1:] # Quitamos el + inicial del primer término
        
    for term in terms[1:]:
        if term.startswith("+"):
            result += f" + {term[1:]}"
        elif term.startswith("-"):
            result += f" - {term[1:]}"
            
    return result

def evaluate(coefs: list, x: float) -> float:
    """Evalúa el polinomio en un valor x dado."""
    if not coefs: return 0.0
    return sum(coefs[i] * (x ** i) for i in range(len(coefs)))

def derivative(coefs: list) -> list:
    """Devuelve los coeficientes del polinomio derivado."""
    if not coefs or len(coefs) == 1:
        return []
    return [coefs[i] * i for i in range(1, len(coefs))]

def main_ex9():
    """Función principal para probar el Ejercicio 9."""
    print("Monomio:", format_monomial(2.0, 3))
    print("Polinomio:", format_polynomial([4.0, 1.0, 0.0, 2.0]))
    print("Evaluación (x=2):", evaluate([4.0, 1.0, 0.0, 2.0], 2))
    print("Derivada:", derivative([4.0, 1.0, 0.0, 2.0]))
```

```python
import re
from datetime import datetime

def procesar_informe(informe: str) -> str:
    """Oculta datos sensibles de un informe médico mediante expresiones regulares."""
    fecha_regex = re.compile(r'\b(\d{1,2})/(\d{1,2})/(\d{4})\b')
    
    # 1. Procesar fecha de nacimiento y buscar la fecha de la consulta
    fechas = fecha_regex.findall(informe)
    # Suponemos que la primera fecha encontrada suele ser la de nacimiento en este formato, 
    # pero buscaremos las líneas específicas
    lineas = informe.split('\n')
    
    fecha_nacimiento = None
    fecha_consulta = None
    
    for linea in lineas:
        if linea.startswith("Fecha de nacimiento:"):
            match = fecha_regex.search(linea)
            if match:
                fecha_nacimiento = datetime.strptime(match.group(0), "%d/%m/%Y")
        elif "Dr." in linea or "Dra." in linea:
            match = fecha_regex.search(linea)
            if match:
                fecha_consulta = datetime.strptime(match.group(0), "%d/%m/%Y")
                
    # Calcular edad
    edad_str = ""
    if fecha_nacimiento and fecha_consulta:
        edad = fecha_consulta.year - fecha_nacimiento.year
        if (fecha_consulta.month, fecha_consulta.day) < (fecha_nacimiento.month, fecha_nacimiento.day):
            edad -= 1
        edad_str = str(edad)
        
    # Reemplazos en orden
    resultado = []
    for linea in lineas:
        if linea.startswith("Fecha de nacimiento:"):
            resultado.append(f"Edad: {edad_str}")
        elif linea.startswith("Nombre:"):
            resultado.append("Nombre: PACIENTE")
        elif "Dr." in linea or "Dra." in linea:
            # Reemplazar nombre médico y la fecha de la consulta
            linea_mod = re.sub(r'(Dr\.|Dra\.)\s+.*', 'MEDICO', linea)
            linea_mod = fecha_regex.sub('FECHA', linea_mod)
            resultado.append(linea_mod)
        else:
            # Reemplazar cualquier otra fecha
            resultado.append(fecha_regex.sub('FECHA', linea))
            
    return "\n".join(resultado)

def main_ex11():
    """Función principal para probar el Ejercicio 11."""
    informe = """Informe clínico de Urgencias
Nombre: Juan Pérez López
Género: Hombre
Fecha de nacimiento: 12/05/1980
Motivo de consulta: dolor abdominal
12/05/2023 Dr. Ramírez
Tratamiento: reposo y analgésicos"""
    
    print(procesar_informe(informe))
```

```python
def is_anagram(word1: str, word2: str) -> bool:
    """Devuelve True si las dos cadenas son anagramas, ignorando mayúsculas y espacios."""
    w1 = word1.lower().replace(" ", "")
    w2 = word2.lower().replace(" ", "")
    return sorted(w1) == sorted(w2)

def group_anagrams(words: list) -> list:
    """Agrupa una lista de palabras en sublistas de anagramas."""
    groups = []
    for word in words:
        found = False
        for group in groups:
            if is_anagram(word, group[0]):
                group.append(word)
                found = True
                break
        if not found:
            groups.append([word])
    return groups

def largest_group(groups: list) -> list:
    """Devuelve el grupo de anagramas con más palabras."""
    if not groups:
        return []
    return max(groups, key=len)

def find_anagrams(word: str, words: list) -> list:
    """Encuentra las palabras en una lista que son anagramas de la palabra dada, excluyendo la palabra en sí."""
    word_lower = word.lower()
    return [w for w in words if is_anagram(w, word) and w.lower() != word_lower]

def main_ex12():
    """Función principal para probar el Ejercicio 12."""
    words = ["roma", "amor", "pelo", "mora", "casa", "lope", "ramo"]
    print("Es anagrama ('Roma', 'amor'):", is_anagram("Roma", "amor"))
    print("Grupos:", group_anagrams(words))
    print("Grupo más grande:", largest_group(group_anagrams(words)))
    print("Encontrar anagramas de 'amor':", find_anagrams("amor", words))
```
